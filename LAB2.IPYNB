## 25. Mini Project: Dataset Health Check Function (Task 24)


```python
# Task 24: Reusable Dataset Health Check Function
def inspect_dataset(df):
    print('================ DATASET HEALTH CHECK ================')
    print('Rows:', df.shape[0])
    print('Columns:', df.shape[1])
    print('Column names:', list(df.columns))
    print('\nMissing values:')
    print(df.isnull().sum())
    print('\nDuplicate rows:')
    print(df.duplicated().sum())
    print('\nData types:')
    print(df.dtypes)
    print('\nFirst 5 rows:')
    print(df.head())
    print('======================================================')

# Run health check on Iris dataset
inspect_dataset(df)
```

- **What does `shape[0]` mean?** Number of rows (150 Iris flower instances).
- **What does `shape[1]` mean?** Number of columns (5 variables: 4 features + 1 target).
