## 7. Activity 2 — Data Cleaning

### 7.1 Finding Missing Values


```python
# Task 11: Identify missing values in each column
print("--- Task 11: Missing Value Counts per Column ---")
print(df.isna().sum())
```

### 7.2 Handling Missing GPA


```python
# Task 12: Calculate mean GPA and fill missing GPA
mean_gpa = df["GPA"].mean()
print(f"Calculated Mean GPA: {mean_gpa:.4f}")
df["GPA"] = df["GPA"].fillna(mean_gpa)

# Task 13: Verify GPA no longer contains missing values
print("\nTask 13: Verification of missing values in GPA:", df["GPA"].isna().sum())
```

### 7.3 Handling Missing Age


```python
# Task 14: Fill missing Age using median age
median_age = df["Age"].median()
print(f"Calculated Median Age: {median_age:.1f}")
df["Age"] = df["Age"].fillna(median_age)

print("Verification of missing values in Age:", df["Age"].isna().sum())
```

**Question:** Why might median be preferable to mean when a variable contains extreme values?  
**Answer:** The median is a resistant measure of central location (it represents the middle value of sorted data) and is not sensitive to extreme outliers. Conversely, the arithmetic mean incorporates every value into its calculation, so extreme low or high values (outliers) heavily skew the mean, making it unrepresentative of the typical value in skewed distributions.


### 7.4 Finding Duplicates


```python
# Task 15: Identify and remove duplicate records
print("Duplicates checking all columns (including StudentID):", df.duplicated().sum())

# Note: Student 111 (Ali) has duplicate student details as Student 101 except for StudentID.
# Checking duplicate records ignoring StudentID:
subset_cols = ["Name", "Age", "Department", "GPA", "Fees"]
print("Duplicate student records (ignoring StudentID):", df.duplicated(subset=subset_cols).sum())

# Remove duplicate records
df = df.drop_duplicates(subset=subset_cols).reset_index(drop=True)
print("Shape after removing duplicate student records:", df.shape)
print("\nCleaned DataFrame after duplicate removal:")
print(df)
```

### 7.5 Checking Category Consistency


```python
print("--- Unique Department values before cleaning ---")
print(df["Department"].unique())

print("\n--- Value counts before cleaning ---")
print(df["Department"].value_counts())

# Task 16: Standardize 'cs' as 'CS' and 'Computer Science' as 'CS'
df["Department"] = df["Department"].replace({
    "cs": "CS",
    "Computer Science": "CS"
})

# Task 17: Run value_counts() again to verify
print("\n--- Task 17: Department Value Counts After Standardization ---")
print(df["Department"].value_counts())
```

## 8. Activity 3 — Grouping and Aggregation

### 8.1 Basic GroupBy


```python
# Task 18: Calculate average GPA for each department
avg_gpa_dept = df.groupby("Department")["GPA"].mean()
print("--- Task 18: Average GPA by Department ---")
print(avg_gpa_dept)
```

```python
# Task 19: Calculate the number of students in each department
student_count_dept = df.groupby("Department")["StudentID"].count()
print("--- Task 19: Student Count by Department ---")
print(student_count_dept)
```

```python
# Task 20: Calculate total fees collected from each department
total_fees_dept = df.groupby("Department")["Fees"].sum()
print("--- Task 20: Total Fees Collected by Department ---")
print(total_fees_dept)
```

### 8.2 Multiple Aggregations


```python
# Task 21: For each department, calculate mean, min, and max GPA
gpa_agg = df.groupby("Department")["GPA"].agg(["mean", "min", "max"])
print("--- Task 21: GPA Aggregations (mean, min, max) by Department ---")
print(gpa_agg)
```

```python
# Task 22: For each department, calculate average Age and average GPA
age_gpa_avg = df.groupby("Department")[["Age", "GPA"]].mean()
print("--- Task 22: Average Age and Average GPA by Department ---")
print(age_gpa_avg)
```

### 8.3 Sorting Grouped Results


```python
# Task 23: Sort departments by average GPA from highest to lowest
sorted_gpa_dept = df.groupby("Department")["GPA"].mean().sort_values(ascending=False)
print("--- Task 23: Departments Sorted by Average GPA (Descending) ---")
print(sorted_gpa_dept)
```

## 9. Integrated Challenge — Clean and Analyze
In this section, we re-execute the complete end-to-end data processing workflow from a raw loading step to final summary stats and write key conclusions.


```python
# Step 1: Load raw data
raw_df = pd.read_csv("students.csv")

# Step 2: Inspect raw data
print("Raw Shape:", raw_df.shape)
print("Missing values per column:\n", raw_df.isna().sum())

# Step 3: Remove duplicate records
cleaned_df = raw_df.drop_duplicates(subset=["Name", "Age", "Department", "GPA", "Fees"]).copy()

# Step 4: Fill missing Age using median
median_age = cleaned_df["Age"].median()
cleaned_df["Age"] = cleaned_df["Age"].fillna(median_age)

# Step 5: Fill missing GPA using mean
mean_gpa = cleaned_df["GPA"].mean()
cleaned_df["GPA"] = cleaned_df["GPA"].fillna(mean_gpa)

# Step 6: Standardize department names
cleaned_df["Department"] = cleaned_df["Department"].replace({
    "cs": "CS",
    "Computer Science": "CS"
})

# Step 7: Find students with GPA >= 3.5
high_achievers = cleaned_df[cleaned_df["GPA"] >= 3.5]
print("\n--- Students with GPA >= 3.5 ---")
print(high_achievers[["StudentID", "Name", "Department", "GPA"]])

# Step 8: Calculate student count by department
dept_counts = cleaned_df.groupby("Department")["StudentID"].count()

# Step 9: Calculate average GPA by department
dept_avg_gpa = cleaned_df.groupby("Department")["GPA"].mean()

# Step 10: Calculate total fees by department
dept_total_fees = cleaned_df.groupby("Department")["Fees"].sum()

# Step 11: Combine and Sort departments by average GPA
dept_summary = pd.DataFrame({
    "Student_Count": dept_counts,
    "Average_GPA": dept_avg_gpa,
    "Total_Fees": dept_total_fees
}).sort_values(by="Average_GPA", ascending=False)

print("\n--- Final Department Summary Table (Sorted by Average GPA) ---")
print(dept_summary)
```

## 12. Short Viva / Discussion Questions

### Question 1: What is the difference between `loc[]` and `iloc[]`?
**Answer:**
- `loc[]` is **label-based**. It accesses rows and columns using explicit index labels (e.g., index label `4`, `'Department'`) and includes the stop boundary when slicing (`0:4` includes label `4`).
- `iloc[]` is **integer position-based**. It accesses elements using 0-indexed positions (e.g., row `0`, column `2`) regardless of label names, and excludes the stop boundary when slicing (`0:4` includes positions `0, 1, 2, 3`).

---

### Question 2: Why can `loc[2]` and `iloc[2]` return different rows?
**Answer:**
`loc[2]` looks for a row whose **index label** is literally `2`. If the DataFrame's index has been re-indexed, filtered, sorted, or set to custom strings (e.g. `['a', 'b', 'c']`), index label `2` might refer to a non-third row or might not exist at all.  
`iloc[2]` always refers strictly to the **3rd physical row** (0-indexed position 2) in the DataFrame, regardless of index label names.

---

### Question 3: What is the difference between `dropna()` and `fillna()`?
**Answer:**
- `dropna()` **removes/deletes** rows or columns containing missing values (`NaN` / `None`) from the DataFrame.
- `fillna()` **replaces/imputes** missing values (`NaN`) with a specified constant value, summary statistic (e.g., mean, median, mode), or propagated values (`ffill`/`bfill`), keeping all rows intact.

---

### Question 4: Why should duplicate records be checked before analysis?
**Answer:**
Duplicate records introduce artificial data inflation. They bias summary statistics (e.g., artificially skewing means, counts, and sums), distort machine learning training distributions, cause overcounting in financial metrics (like total fees), and lead to inaccurate business conclusions. Checking and dropping duplicates ensures data integrity.

---

### Question 5: What does `value_counts()` tell us?
**Answer:**
`value_counts()` returns a Pandas Series containing the frequency counts of unique values present in a Series or DataFrame column, sorted in descending order by default. It helps identify common categories, rare values, dominant classes, and inconsistent categorical entries (e.g., `cs` vs `CS`).

---

### Question 6: What does `groupby()` do?
**Answer:**
`groupby()` implements the **Split-Apply-Combine** strategy:
1. **Split:** Divides the DataFrame into subset groups based on unique values in specified key columns.
2. **Apply:** Applies an aggregation, transformation, or filtering function independently to each group.
3. **Combine:** Merges the group results into a single output data structure (Series or DataFrame).

---

### Question 7: What is the difference between `mean()`, `count()`, and `sum()` after `groupby()`?
**Answer:**
- `mean()`: Computes the arithmetic average of numeric values within each group.
- `count()`: Counts the number of non-null values present within each group.
- `sum()`: Calculates the total numerical sum of values within each group.

---

### Question 8: Why is data cleaning performed before final analysis?
**Answer:**
Data cleaning ensures the accuracy, quality, and reliability of analytical results ("Garbage In, Garbage Out"). Raw data often contains missing values, duplicates, formatting inconsistencies, and invalid inputs. Performing analysis on uncleaned data leads to invalid group counts, distorted averages, broken queries, and flawed decision-making.
